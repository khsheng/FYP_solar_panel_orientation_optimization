# Dataset Structure

## Function inspect_csv

```python
from pathlib import Path

import pandas as pd

DATA_ROOT = next(
    path for path in (Path("dataset/row"), Path("../dataset/row"))
    if path.is_dir()
)


def inspect_csv(path, sample_size=5, chunk_size=100_000, skiprows=0):
    row_count = 0
    column_names = None
    sample_records = None
    dtype_observations = {}

    for chunk in pd.read_csv(path, skiprows=skiprows, chunksize=chunk_size):
        if column_names is None:
            column_names = chunk.columns.tolist()
            sample_records = chunk.head(sample_size)
            dtype_observations = {column: set() for column in column_names}

        row_count += len(chunk)
        for column, dtype in chunk.dtypes.items():
            dtype_observations[column].add(str(dtype))

    print(f"File: {path}")
    print(f"Rows: {row_count:,}")
    print(f"Columns: {len(column_names)}")
    print("\nColumn names:")
    print(column_names)
    print("\nData types:")
    for column in column_names:
        observed_types = sorted(dtype_observations[column])
        dtype_summary = observed_types[0] if len(observed_types) == 1 else (
            f"mixed across chunks ({', '.join(observed_types)})"
        )
        print(f"  {column}: {dtype_summary}")
    print("\nSample records:")
    display(sample_records)
```

## Temporal Analysis

Timestamps are normalized to Malaysia Time (Asia/Kuala_Lumpur, UTC+08:00) for comparison; source CSVs are not modified. For each dataset, the available-data interval summary uses consecutive unique, valid timestamps. The normal/nominal cadence is the most common interval (mode), with its rate and share of observed intervals reported. The available-data rate is the average number of observed intervals per elapsed day across the full first-to-last timestamp span.

### Shared temporal analysis helper

```python
MALAYSIA_TZ = "Asia/Kuala_Lumpur"


def report_temporal_analysis(dataset_name, timestamps, source_timezone):
    timestamps = pd.Series(timestamps)
    invalid_count = int(timestamps.isna().sum())
    valid_timestamps = timestamps.dropna()
    duplicate_count = int(valid_timestamps.duplicated().sum())
    unique_timestamps = (
        valid_timestamps.drop_duplicates().sort_values().reset_index(drop=True)
    )

    if len(unique_timestamps) < 2:
        raise ValueError(f"{dataset_name} needs at least two valid unique timestamps")

    intervals = unique_timestamps.diff().dropna()
    nominal_interval = intervals.mode().iloc[0]
    interval_days = (unique_timestamps.iloc[-1] - unique_timestamps.iloc[0]).total_seconds() / 86_400
    available_rate = len(intervals) / interval_days
    nominal_rate = 86_400 / nominal_interval.total_seconds()
    nominal_share = 100 * (intervals == nominal_interval).mean()

    print(f"Dataset: {dataset_name}")
    print(f"Source timezone: {source_timezone}")
    print(f"Analysis timezone: {MALAYSIA_TZ} (UTC+08:00)")
    print(f"Earliest timestamp: {unique_timestamps.iloc[0].strftime('%Y-%m-%d %H:%M:%S')} UTC+08:00")
    print(f"Latest timestamp:   {unique_timestamps.iloc[-1].strftime('%Y-%m-%d %H:%M:%S')} UTC+08:00")
    print(f"Valid observations: {len(valid_timestamps):,}; unique timestamps: {len(unique_timestamps):,}")
    print(f"Invalid timestamps: {invalid_count:,}; duplicate timestamps: {duplicate_count:,}")
    print(f"Available-data intervals (min / median / max): {intervals.min()} / {intervals.median()} / {intervals.max()}")
    print(f"Available-data rate over the full timestamp span: {available_rate:.2f} intervals/day")
    print(f"Normal/nominal cadence (most common interval): every {nominal_interval}")
    print(f"Normal/nominal frequency: {nominal_rate:.2f} observations/day; {nominal_share:.2f}% of observed intervals match")
```

### Himawari-8 CLP 2016-2020

The `timestamp` column is UTC. Convert it to Malaysia Time and verify the conversion against the dataset's existing `malaysia_datetime` column.

```python
path = DATA_ROOT / "himawari8" / "CLP" / "CLP_2016_2020.csv"
frame = pd.read_csv(path, usecols=["timestamp", "malaysia_datetime"])
timestamps = pd.to_datetime(frame["timestamp"], utc=True).dt.tz_convert(MALAYSIA_TZ)
recorded_malaysia_time = pd.to_datetime(frame["malaysia_datetime"]).dt.tz_localize(MALAYSIA_TZ)
assert timestamps.equals(recorded_malaysia_time), "Converted timestamps do not match malaysia_datetime"
report_temporal_analysis("Himawari-8 CLP 2016-2020", timestamps, "UTC (timestamp column)")
```

### NSRDB 2016-2020

The NSRDB metadata identifies the source `Time Zone` as 0 (UTC). Convert the year/month/day/hour/minute fields to Malaysia Time and verify them against the existing `Malaysia Datetime` column.

```python
path = DATA_ROOT / "NSRDB" / "NSRDB_2016_2020_combined.csv"
frame = pd.read_csv(
    path,
    skiprows=2,
    usecols=["Year", "Month", "Day", "Hour", "Minute", "Malaysia Datetime"],
)
utc_timestamps = pd.to_datetime(
    frame[["Year", "Month", "Day", "Hour", "Minute"]].rename(
        columns={"Year": "year", "Month": "month", "Day": "day", "Hour": "hour", "Minute": "minute"}
    )
).dt.tz_localize("UTC")
timestamps = utc_timestamps.dt.tz_convert(MALAYSIA_TZ)
recorded_malaysia_time = pd.to_datetime(frame["Malaysia Datetime"]).dt.tz_localize(MALAYSIA_TZ)
assert timestamps.equals(recorded_malaysia_time), "Converted timestamps do not match Malaysia Datetime"
report_temporal_analysis(
    "NSRDB 2016-2020",
    timestamps,
    "UTC (Time Zone=0 in NSRDB metadata)",
)
```

# Statistical Analysis

For every numerical column, calculate the non-missing count, mean, median, sample standard deviation, minimum, maximum, zero count, negative-value count, and missing-value count. CSVs are processed in chunks; a temporary disk-backed array is used to calculate exact medians without loading the full wide NSRDB dataset into memory. Temporary files are removed after each dataset.

## Shared statistical analysis helper

```python
import tempfile

import numpy as np


def summarize_numeric_columns(dataset_name, path, skiprows=0, chunk_size=5_000):
    row_count = 0
    numeric_columns = None
    temp_path = None

    with tempfile.TemporaryDirectory(prefix="dataset_stats_") as temp_dir:
        temp_path = Path(temp_dir) / "numeric_values.bin"
        with temp_path.open("wb") as temp_file:
            for chunk in pd.read_csv(path, skiprows=skiprows, chunksize=chunk_size):
                numeric_chunk = chunk.select_dtypes(include="number")
                if numeric_columns is None:
                    numeric_columns = numeric_chunk.columns.tolist()
                    if not numeric_columns:
                        raise ValueError(f"No numerical columns found in {path}")
                elif numeric_chunk.columns.tolist() != numeric_columns:
                    raise ValueError(f"Numerical columns changed between chunks in {path}")

                numeric_chunk.to_numpy(dtype=np.float64, na_value=np.nan).tofile(temp_file)
                row_count += len(chunk)

        if row_count == 0:
            raise ValueError(f"No data rows found in {path}")

        numeric_values = np.memmap(
            temp_path,
            dtype=np.float64,
            mode="r",
            shape=(row_count, len(numeric_columns)),
        )
        try:
            records = []
            for column_index, column_name in enumerate(numeric_columns):
                values = pd.Series(numeric_values[:, column_index], copy=False)
                records.append(
                    {
                        "column": column_name,
                        "count": int(values.count()),
                        "mean": values.mean(),
                        "median": values.median(),
                        "std": values.std(),
                        "min": values.min(),
                        "max": values.max(),
                        "zero_count": int(values.eq(0).sum()),
                        "negative_count": int(values.lt(0).sum()),
                        "missing_count": int(values.isna().sum()),
                    }
                )
                del values
        finally:
            del numeric_values

    summary = pd.DataFrame.from_records(records).set_index("column")
    print(f"{dataset_name}: {row_count:,} rows; {len(numeric_columns):,} numerical columns")
    display(summary)
    return summary
```

## Dataset: Himawari-8 CLP 2016-2020

Summary statistics for all numeric columns in the 2016-2020 Himawari-8 CLP dataset.

```python
himawari_2016_2020_stats = summarize_numeric_columns(
    "Himawari-8 CLP 2016-2020",
    DATA_ROOT / "himawari8" / "CLP" / "CLP_2016_2020.csv",
)
```
